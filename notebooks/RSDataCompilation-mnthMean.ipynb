```python
import pandas as pd
import os
import numpy as np
```

```python
base = os.path.join(os.getcwd(),'..')
```

```python
stats = os.path.join(base,'stats')
```

```python
def get_df_gpp(fn):
    df = pd.read_csv(fn)
    df = df[['State_Name', 'month', 'mean', 'year']]
    df.dropna(inplace=True)
    df['month']=df['month'].astype(int)
    return df
```

```python
gpp_fol = os.path.join(stats,'rs data','Gpp_values_mean')
```

```python
compiled = pd.DataFrame()
for r,d,f in os.walk(gpp_fol):
    for fl in f:
        fn = os.path.join(r,fl)
        compiled = pd.concat([compiled,get_df_gpp(fn)])
```

```python
compiled.to_excel(os.path.join(base,'compiled values','rs_gpp_compiled.xlsx'))
```
