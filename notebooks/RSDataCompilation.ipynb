```python
import pandas as pd
import os
import numpy as np
```

```python
base = os.path.join(os.getcwd(),'..')
```

```python
stats = os.path.join(base,'stats')
```

```python
hdate = pd.read_csv(os.path.join(stats,'rs data','Maize_Harvest_MultiYear_Derivative_States_07-20.csv'))
df = hdate[['State_Name','mean','year']]
df=df.dropna()
df['mean']=round(df['mean'].astype(int))
df['year']=round(df['year'].astype(int))
df.to_excel(os.path.join(base,'compiled values','rs_hdate_clean.xlsx'))
```

```python
sdate = pd.read_csv(os.path.join(stats,'rs data','Maize_SOS_MultiYear_Derivative_States_07-20.csv'))
df = sdate[['State_Name','mean','year']]
df=df.dropna()
df['mean']=round(df['mean'].astype(int))
df['year']=round(df['year'].astype(int))
df.to_excel(os.path.join(base,'compiled values','rs_sdate_clean.xlsx'))
```
