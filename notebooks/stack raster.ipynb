```python
import rasterio as rio
import os
import matplotlib.pyplot as plt
```

```python
base = os.path.join(os.getcwd(),'..','raster(clip)','aboveground_biomassc')
```

```python
ls = []
for r,d,f in os.walk(base):
    for fl in f:
        if 'irrigated' in fl:
            ls.append(os.path.join(r,fl))
```

```python
fo = os.path.join(base,'AGBM_irrigated.tif')
```

```python
with rio.open(ls[0]) as src0:
    meta = src0.meta

meta.update(count=len(ls))

with rio.open(fo, "w", **meta) as dst:
    for i, layer in enumerate(ls, start=1):
        with rio.open(layer) as src:
            dst.write(src.read(1), i)
```
