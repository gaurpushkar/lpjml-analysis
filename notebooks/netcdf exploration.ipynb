```python
base = os.path.join(os.getcwd(),'..')
```

```python
data = os.path.join(base,'data(LPJmL)','simulation_global_gwwd2_pushkar')
```

```python
fn = os.path.join(data,'cft_evap.nc')
```

```python
ds = nc.Dataset(fn)
```

```python
ds
```

```python
def read_nc_params(fn):
    name=[]
    size=[]
    for dim in ds.dimensions.values():
        name.append(dim.name)
        size.append(dim.size)
    dt = pd.DataFrame()
    dt['name']=name
    dt['size']=size
    dt = dt.set_index('name')
    dt=dt.T
    return dt
```

```python
df = pd.DataFrame()
for r,d,f in os.walk(data):
    for fl in f:
        if fl.endswith('.nc'):
            fn = os.path.join(r,fl)
            df= pd.concat([df,read_nc_params(fn)])
```
