```python
import pandas as pd
import os
import numpy as np
```

```python
base = os.path.join(os.getcwd(),'..')
```

```python
inDir = os.path.join(base,'stats')
outDir = os.path.join(base,'compiled values')
```

```python
var_name= 'aboveground_biomassc'
cult = 'rainfed'
cVals = os.path.join(inDir,var_name,var_name+'_'+cult+'_compiled.xlsx')
```

```python
df = pd.read_excel(cVals)
```

```python
test = pd.DataFrame()
for d in range(0,10):
    dt= df[['State_Name',d]]
    dt['time']=d
    dt.rename(columns={d:'Mean(SO)'},inplace=True)
    test = pd.concat([test,dt],axis=0)
# test['date']=test['time'].apply(lambda x:pd.to_datetime(2010,format='%Y')+pd.DateOffset(months=x))
test['date'] = test['time'].apply(lambda x:pd.to_datetime(2010+x,format='%Y'))

test.drop(columns=['time'],inplace=True)
```

```python
test.to_excel(os.path.join(outDir,var_name+'_'+cult+'_compiled.xlsx'))
```
