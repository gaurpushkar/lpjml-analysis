```python
import pandas as pd
import os
```

```python
base = os.path.join(os.getcwd(),'..','compiled values')
```

```python
def prep_df(df,col):
    df['idate'] = df['Date'].apply(lambda x:str(x.year)+'-'+str(x.month))
    df['index'] = df['State']+'-'+df['idate']
    df = df.set_index('index')
    df.drop(columns=['Unnamed: 0','State','Date','idate'],inplace=True)
    df.rename(columns={'max':col},inplace=True)
    return df
```

```python
leaf_ir = pd.read_excel(os.path.join(base,'Compiled_leaf_irrigated.xlsx'))
leaf_rf = pd.read_excel(os.path.join(base,'Compiled_leaf_rainfed.xlsx'))

pool_ir = pd.read_excel(os.path.join(base,'Compiled_pool_irrigated.xlsx'))
pool_rf = pd.read_excel(os.path.join(base,'Compiled_pool_rainfed.xlsx'))

so_ir = pd.read_excel(os.path.join(base,'Compiled_so_irrigated.xlsx'))
so_rf = pd.read_excel(os.path.join(base,'Compiled_so_rainfed.xlsx'))
```

```python
leaf_ir = prep_df(leaf_ir,'leaf')
leaf_rf = prep_df(leaf_rf,'leaf')

pool_ir = prep_df(pool_ir,'pool')
pool_rf = prep_df(pool_rf,'pool')

so_ir = prep_df(so_ir,'so')
so_rf = prep_df(so_rf,'so')
```

```python
agbm_ir = pd.concat([leaf_ir,pool_ir,so_ir],axis=1)
agbm_ir = agbm_ir.sum(axis=1)
agbm_ir = pd.DataFrame(agbm_ir)
agbm_ir['id'] = agbm_ir.index
agbm_ir['State'] = agbm_ir['id'].apply(lambda x:x.split('-')[0])
agbm_ir['year'] = agbm_ir['id'].apply(lambda x:x.split('-')[1])
agbm_ir['month'] = agbm_ir['id'].apply(lambda x:x.split('-')[2])
agbm_ir.drop(columns=['id'],inplace=True)
agbm_ir.rename(columns={0:'AGBM(IR)'},inplace=True)
```

```python
agbm_ir.to_excel(os.path.join(base,'Compiled_agbm(ir).xlsx'))
```

```python
agbm_rf = pd.concat([leaf_rf,pool_rf,so_rf],axis=1)
agbm_rf = agbm_rf.sum(axis=1)
agbm_rf = pd.DataFrame(agbm_rf)
agbm_rf['id'] = agbm_rf.index
agbm_rf['State'] = agbm_rf['id'].apply(lambda x:x.split('-')[0])
agbm_rf['year'] = agbm_rf['id'].apply(lambda x:x.split('-')[1])
agbm_rf['month'] = agbm_rf['id'].apply(lambda x:x.split('-')[2])
agbm_rf.drop(columns=['id'],inplace=True)
agbm_rf.rename(columns={0:'AGBM(RF)'},inplace=True)
```

```python
agbm_rf.to_excel(os.path.join(base,'Compiled_agbm(rf).xlsx'))
```

```python

so_ir['id'] = so_ir.index
so_ir['State'] = so_ir['id'].apply(lambda x:x.split('-')[0])
so_ir['year'] = so_ir['id'].apply(lambda x:x.split('-')[1])
so_ir['month'] = so_ir['id'].apply(lambda x:x.split('-')[2])
so_ir.drop(columns=['id'],inplace=True)
so_ir.rename(columns={0:'Yield(IR)'},inplace=True)
```

```python

so_rf['id'] = so_rf.index
so_rf['State'] = so_rf['id'].apply(lambda x:x.split('-')[0])
so_rf['year'] = so_rf['id'].apply(lambda x:x.split('-')[1])
so_rf['month'] = so_rf['id'].apply(lambda x:x.split('-')[2])
so_rf.drop(columns=['id'],inplace=True)
so_rf.rename(columns={0:'so(rf)'},inplace=True)
```

```python
so_ir.to_excel(os.path.join(base,'Compiled_so(ir).xlsx'))
so_rf.to_excel(os.path.join(base,'Compiled_so(rf).xlsx'))
```
