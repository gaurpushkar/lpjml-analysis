```python
import pandas as pd
import os
import numpy as np
```

```python
base = os.path.join(os.getcwd(),'..')
sts = os.path.join(base,'stats')
outPath = os.path.join(base,'compiled values')
```

```python
def get_df(fn,t):
    df=pd.read_csv(fn)
    df = df[['zone_id','max']]
    df.rename(columns={'zone_id':'State'},inplace=True)
    df['Date']=t
    return df
```

```python
comp = pd.DataFrame()
sd = 2015
param = 'sdate'
cond = 'irrigated'
for r,d,f in os.walk(sts):
    for fl in f:
        if (param in fl) and (cond in fl):
            fn = os.path.join(r,fl)
            print(fl)
            t = int(fl.split('_')[2])
            # time = pd.to_datetime(sd,format='%Y')+pd.DateOffset(months=t)
            time = sd+t
            comp = pd.concat([comp,get_df(fn,time)])
comp.to_excel(os.path.join(outPath,'Compiled_'+param+'_'+cond+'.xlsx'))
```
