```python
import os
import rasterio
from rasterio.mask import mask
import geopandas as gpd
import numpy as np
```

```python
base = os.path.join(os.getcwd(),'..')
```

```python
shapefile_path = os.path.join(os.getcwd(),'..','shape','States_shapefile.shp')
```

```python
outPath = os.path.join(base,'raster(clip)')
if not os.path.exists(outPath):
    os.mkdir(outPath)
```

```python
def save_raster(fn,fo,band_number):
    gdf = gpd.read_file(shapefile_path)
    with rasterio.open(fn) as src:
        if gdf.crs != src.crs:
            gdf = gdf.to_crs(src.crs)
    
        geometries = gdf.geometry.values
        clipped_image, clipped_transform = mask(
            src,
            geometries,
            crop=True,
            nodata = 0,
            indexes=band_number 
        )
        clipped_meta = src.meta.copy()
        clipped_meta.update({
            "driver": "GTiff",
            "height": clipped_image.shape[0],
            "width": clipped_image.shape[1],
            "transform": clipped_transform,
            "count": 1 
        })
    
    with rasterio.open(fo, "w", **clipped_meta) as dest:
        dest.write(clipped_image, 1)

    print('Output in folder',fo)
```

```python
def get_3(n):
    x = len(str(n))
    y = '0'*(3-x)+str(n)
    return y
```

```python
var_name = 'SWC'
bn = [1]
ir_type = ['rainfed','irrigated']
os.makedirs(os.path.join(base,'raster(clip)',var_name),exist_ok=True)
for i in range(len(bn)):
    for t in range(0,120):
        ts = get_3(t)
        fn = os.path.join(base,'tiffs',var_name,var_name+'_time_'+str(ts)+'.tif')
        fo = os.path.join(base,'raster(clip)',var_name,var_name+'_time_'+str(t)+'_'+ir_type[i]+'.tif')
        save_raster(fn,fo,bn[i])
```
