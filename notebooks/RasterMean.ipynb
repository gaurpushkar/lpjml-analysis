```python
import rasterio
import numpy as np
import os
```

```python
base = os.path.join(os.getcwd(),'..','raster(clip)','aboveground_biomassc')
```

```python
input_file = os.path.join(base,'AGBM_rainfed.tif')
output_file = os.path.join(base,'AGBM_rainfed_mean.tif')

```

```python
with rasterio.open(input_file) as src:
    data = src.read()
    
    if src.nodata is not None:
        data = np.where(data == src.nodata, np.nan, data)
    
    mean_raster = np.nanmean(data, axis=0)
    
    meta = src.meta.copy()
    meta.update(count=1, dtype='float32')

    with rasterio.open(output_file, "w", **meta) as dst:
        dst.write(mean_raster.astype('float32'), 1)
```
